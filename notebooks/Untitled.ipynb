```python
from pathlib import Path
import h5py
from anndata.io import read_elem

folder = Path("/home/vivek/NOBINFBACKUP/projects/mGast_tchic/adata_files")

for name in [
    "allsamples_genelevel_RNAandChIC_counts.h5ad",
    "allsamples_genelevel_RNAandChIC_counts.filt-norm.h5ad",
]:
    with h5py.File(folder / name, "r") as f:
        obs = read_elem(f["obs"])
        embeddings = list(f["obsm"]) if "obsm" in f else []

        print(f"\n{name}")
        print("Metadata-kolommen:", list(obs.columns))
        print("Embeddings:", embeddings)
        print("UMAP aanwezig:", any("umap" in key.lower() for key in embeddings))

        for column in obs.columns:
            if obs[column].astype("string").str.contains(
                "endoderm", case=False, na=False
            ).any():
                print(f"Endoderm-labels in '{column}':")
                print(obs[column].value_counts())
```

```python
from pathlib import Path
import h5py

folder = Path("/home/vivek/NOBINFBACKUP/projects/mGast_tchic/adata_files")

def show_entry(name, obj):
    if isinstance(obj, h5py.Dataset):
        print(f"  {name}: shape={obj.shape}, dtype={obj.dtype}")
    else:
        print(f"  {name}/")

for filename in [
    "allsamples_genelevel_RNAandChIC_counts.h5ad",
    "allsamples_genelevel_RNAandChIC_counts.filt-norm.h5ad",
]:
    print(f"\n--- {filename} ---")
    with h5py.File(folder / filename, "r") as f:
        f.visititems(show_entry)
```
